## Problem and data

Where the data comes from, its size, and the train/test split.

```python
n, d = 100, 20
A = rng.normal(size=(n, d))
x_true = rng.normal(size=d)
b = A @ x_true + 0.1 * rng.normal(size=n)
print(f"A: {A.shape}, b: {b.shape}")
```

## Step 1: Method name

**Idea.** One paragraph in plain words.

**Update rule.** $x^{(k+1)} = x^{(k)} - t\,\nabla f(x^{(k)})$

```python
def f(x, lam=0.1):
    return 0.5 * np.sum((A @ x - b) ** 2) + lam * np.sum(x ** 2)

def grad(x, lam=0.1):
    return A.T @ (A @ x - b) + 2 * lam * x

t = 1.0 / np.linalg.norm(A, 2) ** 2
x = np.zeros(d)
history = []
for k in range(500):
    x = x - t * grad(x)
    history.append(f(x))
print(f"final objective: {history[-1]:.6f}")
```

```python
fig, ax = plt.subplots(figsize=(6.5, 4))
ax.semilogy(np.array(history) - min(history) + 1e-12, color=COLORS["gd"], label="GD")
ax.set_xlabel("iteration"); ax.set_ylabel("f(x) - f*"); ax.legend()
save(fig, "fig1_convergence")
```
